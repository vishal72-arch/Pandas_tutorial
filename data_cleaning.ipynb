```python
import pandas as pd
```

```python
data = {
    "Image_ID": ["img001.jpg", "img002.jpg", "img003.jpg", "img004.jpg"],
    "Label": ["Fire", "No Fire", "Fire", "No Fire"],
    "Confidence": [0.92, 0.87, 0.95, 0.81],
    "Width": [640, 640, 1280, 1280],
    "Height": [640, 640, 720, 720]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
#how many null values we have
df.isnull().sum()
```

```python
#droping all null values
df.dropna(how = "any")#any row that have null value will be deleted
df.dropna(how = "all")#drop rows if any only if all the values in the row is null
```

```python
#filling null values or missing values using fillna()
df.fillna()#inside parentheses provide the value to fill
#i has 2 methofs forward fill and backward fill
```

```python
df['Image_ID'].replace("img001.jpg" , "new_img")
```

```python
#dealing with duplicate values
df[df.duplicated(keep = 'first')]#leave first element and choose the other duplicate element 
```

```python
#droping duplicates
df = df.drop_duplicates()
#the Whole record gets deleted i.e row
```
