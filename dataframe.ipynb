```python
import pandas as pd
```

```python
data = {
    "Name": ["Vishal", "Rahul", "Priya", "Aman"],
    "Age": [21, 22, 20, 23],
    "City": ["Delhi", "Mumbai", "Pune", "Bangalore"],
    "Salary": [50000, 60000, 55000, 70000]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
#for seeing starting values
df.head(2) #syntax --)df.head(n) where n is number of rows
```

```python
#and same as for tail only difference is tail is used to see last value
df.tail(2)
```

```python
df.iloc[1:3 , :2] #in this we can access certain columns but keep in mind iloc is for location index(numerical) , and loc is for label
#syntax --> rows , column
```

```python
#now we use loc when when are fetching rows 1 or 2 certain columns
df.loc[1:3 , ["Age" , "City"]]
```

```python
# For accessing particular columns in df we have the following syntax
df['Age']#for accessing single column use 1d array and for accessing more than one column use 2d array
```

```python
#for accessing multiple columns we place the column names in a 2d array
df[['Name' , 'Age' , 'City']]
```

```python
#deleting or droping columns
df.drop(0 , axis = 0)

#for deleting rows we need to provide either index label or index name in the following syntax
#df.drop(numerical index or index label , axis = 0)

#for deleting columns we need to provide the column name and axis = 1
#df.drop(column_name , axis = 1)

#inplace parameter mean perform operation on original dataframe and changes will be permanent
```
