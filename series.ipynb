```python
import pandas as pd
```

```python
series_int = pd.Series([10 , 20 ,30 , 40 ,50])
series_int
# In this the data type will be int64 as there is no extra data type other than int
```

```python
series = pd.Series([10 , 20 , 30 , 40 , 'abc'])
series
#now the data type of the series would be object as there is now a single high proiority data type i.e string
```

SERIES INDEXING

ANOTHER WAY OF DOOING INDEXING US ILOC
WHICH IS BASICALLY LOCATION BASED INDEXING

```python
#ILOC --> LOCATION BASED INDEXING
#uses index and fetch the values at that particular indexes
series.iloc[3]
```

```python
#for passing multiple values in iloc
series_int[[1,2,4]]
```

```python
#how to change index values
index = ['cake' , 'banana' , 'chocolate' , 'apple' , 'kiwi']
series_int.index = index
series_int
```

ILOC WAS NUMERICAL BASED INDEXING THAT ACCEPTS ONLY NUMERICAL INDEXES

LOC IS LABEL BASED INDEXING THAT ACCEPTS LABELS I.E STRINGS

```python
series_int.loc['apple']
#In this label based indexing both start and end values are included in the output
series_int['cake':'kiwi']
```
